### Wassertein Distance

```python
import os
import glob
import xarray as xr
import numpy as np
from tqdm import tqdm  # Add tqdm import
from statsmodels.distributions.empirical_distribution import ECDF

# === Paths and Variables ===
# validation_dir = "C:/Users/bmk7346/Documents/Downscale_Proj/Results/2.AORC/1981-2011/validation/"
validation_dir = "C:/Users/bmk7346/Documents/Downscale_Proj/Results/2.AORC/1981-2011/validation_ecdfloss/"
var_name = "pr"

# ensemble_models = ["ACCESS-CM2", "MIROC6", "ACCESS-ESM1-5", "MIROC-ES2L", "MPI-ESM1-2-HR",
#                    "CNRM-ESM2-1", "IPSL-CM6A-LR", "MPI-ESM1-2-LR", "MRI-ESM2-0",
#                    "EC-Earth3-Veg-LR", "NESM3"]

ensemble_models = ["ACCESS-CM2", "MIROC6"]

# === Load Target ===
target_files = sorted(glob.glob(os.path.join(validation_dir, ensemble_models[0], "target*")))
target_ds = xr.concat([xr.open_dataset(f)[var_name] for f in target_files], dim="time")
obs = target_ds.load()  # Shape: [time, lat, lon]

# === Load Ensemble Members with tqdm ===
ensemble_data = []
for model in tqdm(ensemble_models, desc="Loading ensemble models"):
    model_files = sorted(glob.glob(os.path.join(validation_dir, model, "output*")))
    model_ds = xr.concat([xr.open_dataset(f)[var_name] for f in model_files], dim="time")
    ensemble_data.append(model_ds)

# Stack into single array: shape [n_models, time, lat, lon]
ensemble_array = xr.concat(ensemble_data, dim="model")
ensemble_array['model'] = ensemble_models  # assign model names
```

```python
# Check for NaN values and data shape
print("obs shape:", obs.shape)
print("ensemble_data[0] shape:", ensemble_data[0].shape)
print("obs nan count:", np.isnan(obs.values).sum())
print("ensemble_data[0] nan count:", np.isnan(ensemble_data[0].values).sum())
print("obs min/max:", np.nanmin(obs.values), np.nanmax(obs.values))
print("ensemble_data[0] min/max:", np.nanmin(ensemble_data[0].values), np.nanmax(ensemble_data[0].values))

print("Model files for first model:", sorted(glob.glob(os.path.join(validation_dir, ensemble_models[0], "output*"))))
ds = xr.open_dataset(sorted(glob.glob(os.path.join(validation_dir, ensemble_models[0], "output*")))[0])
print("Variables in file:", list(ds.variables))
print("Sample data (first 5 values):", ds[var_name].values.flatten()[:5])

for idx, ens in enumerate(ensemble_data):
    print(f"ensemble_data[{idx}] min/max:", np.nanmin(ens.values), np.nanmax(ens.values))
```

### Library function version, skips the ECDF caluclations

```python
# Calculate the Wasserstein distance for each pixel between each model's ECDF (ecdf_grid) and the observation's ECDF (obs_ecdf_grid)
from scipy.stats import wasserstein_distance

# Assume: ensemble_data is a list of DataArrays [model, time, lat, lon]
n_models = len(ensemble_data)
n_time, n_lat, n_lon = ensemble_data[0].shape

w_distance_grid = np.full((n_models, n_lat, n_lon), np.nan)  # shape: [model, lat, lon]

for m in tqdm(range(n_models), desc="Models"):
    model_ds = ensemble_data[m]  # shape: [time, lat, lon]
    for i in tqdm(range(n_lat), desc=f"Lat (model {m+1})", leave=False):
        for j in range(n_lon):
            model_series = model_ds[:, i, j].values
            obs_series = obs[:, i, j].values
            if np.all(np.isnan(model_series)) or np.all(np.isnan(obs_series)):
                continue  # skip all-nan pixels
            # Compute Wasserstein distance between model and obs time series at this pixel
            w_distance_grid[m, i, j] = wasserstein_distance(model_series[~np.isnan(model_series)], obs_series[~np.isnan(obs_series)])


# Convert to xarray.DataArray with model names as coordinate
w_distance_xr = xr.DataArray(
    w_distance_grid,
    dims=["model", "lat", "lon"],
    coords={
        "model": ensemble_models,
        "lat": np.arange(n_lat),
        "lon": np.arange(n_lon)
    },
    name="wasserstein_distance"
)

np.save("wasserstein_distance_ecdfloss.npy", w_distance_grid)
```

```python
# Plotting the Wasserstein distance grid
import matplotlib.pyplot as plt

# Average over models
avg_w_distance = np.mean(w_distance_grid, axis=0)

# Set colorbar limits: vmin=0, vmax=rounded-up max value
vmax = np.round(np.nanmax(avg_w_distance) + 0.05, 1)

plt.figure(figsize=(10, 6))
plt.imshow(np.flip(avg_w_distance, axis=0), cmap='viridis', vmin=0, vmax=vmax)  # flip on y axis (latitude)    
plt.colorbar(label='Wasserstein Distance')
plt.title('Average Wasserstein Distance between Model Ensembles and Observations')
plt.xlabel('Longitude Index')
plt.ylabel('Latitude Index')
plt.show()
```
